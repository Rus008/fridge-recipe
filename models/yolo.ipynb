```python
from ultralytics import YOLO
import torch
from PIL import Image
import os
from ultralytics.nn.tasks import DetectionModel

# --- Начало обходного пути для PyTorch >= 2.6 ---
# Сохраняем оригинальную функцию torch.load
_original_torch_load = torch.load

# Определяем новую функцию, которая вызывает оригинальную с weights_only=False
def _custom_torch_load(*args, **kwargs):
    kwargs['weights_only'] = False
    return _original_torch_load(*args, **kwargs)

# Временно заменяем torch.load нашей версией
torch.load = _custom_torch_load
# --- Конец обходного пути ---

# Загрузка модели
model = YOLO("yolo11n.pt")

# --- Восстановление оригинальной функции torch.load ---
torch.load = _original_torch_load
# --- Конец восстановления ---

# Обучение модели
# train_results = model.train(
#     data="coco8.yaml",  # путь к YAML датасета
#     epochs=3,  # количество эпох обучения
#     imgsz=640,  # размер изображения для обучения
#     device="cpu",  # устройство для запуска, например, device=0 или device=0,1,2,3 или device=cpu
# )
# 
# # Оценка производительности модели на валидационном наборе
# metrics = model.val()

# Обнаружение объектов на изображении
results = model("../data/fridge.jpg")
im_array = results[0].plot()
im = Image.fromarray(im_array[..., ::-1])  # BGR to RGB
im.show()

# Экспорт модели в формат ONNX
path = model.export(format="onnx")  # возвращает путь к экспортированной модели
```
